## 12

```python
def sum_of_reciprocals(n):
    total_sum = 0
    for i in range(1, n+1):
        total_sum += 1 / i
    return total_sum
n = int(input("Enter a positive integer: "))
if n > 0:
    result = sum_of_reciprocals(n)
    print(f"The sum of reciprocals from 1 to {n} is: {result:.2f}")
else:
    print("Please enter a positive integer.")
```

## 14

```python
def factorial(n):
    if n == 0:
        return 1
    elif n > 0:
        result = 1
        for i in range(1, n + 1):
            result *= i
        return result
    else:
        return None

pos_number = int(input("Enter a positive integer: "))

if pos_number < 0:
    print("Factorial does not exist for negative numbers.")
elif pos_number == 0:
    print("The factorial of 0 is 1.")
else:
    fact = factorial(pos_number)
    print(f"The factorial of {pos_number} is {fact}.")
```

## 15

```python
def power(base, exponent):
    if exponent >= 0:
        result = 1
        for _ in range(exponent):
            result *= base
        return result
    else:
        result = 1
        for _ in range(abs(exponent)):
            result *= base
        return 1 / result
base = float(input("Enter the base number: "))
exponent = int(input("Enter the exponent: "))
result = power(base, exponent)
print(f"{base} raised to the power of {exponent} is {result}")
```
