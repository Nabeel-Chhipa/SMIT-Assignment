```python
import json

deposit_path = 'transactions_history.json'
withdrawal_path = 'withdrawal_history.json'

## CREATE ACCOUNT
def create_account(name = '', balance = 0, transactions = []):
    account = {
        'name': name,
        'balance': balance,
        'transactions': transactions
    }
    print(f'''Create account Data -> Name: {name}, balance: {balance}, transactions: {transactions}''')
    return account

## DEPOSIT MONEY
def deposit_money(account, deposit_amount):
    if deposit_amount > 0:
        account['balance'] += deposit_amount
        account['transactions'].append(deposit_amount)
        deposit_history(account['transactions'])
        message = print(f'User Data from deposit_money function: {account}')
    else:

        print(f'Deposit amount must be grater than 0')
    return message

## SAVE DEPOSIT FILE
def deposit_history(data):
    with open(deposit_path, 'w+') as file:
        json.dump(data, file)
        
## SAVE WITHDRAWAL FILE
def withdraw_history(data):
    with open(withdrawal_path, 'w+') as file:
        json.dump(data, file)

## WITHDRAW MONEY
def withdraw_money(account, withdraw_amount):
    if 'withdrawal' not in account:
        account['withdrawal'] = []
        
    _balance = account['balance']
    if withdraw_amount > 0:
        if withdraw_amount <= _balance:
            account['balance'] -= withdraw_amount
            account['withdrawal'].append(withdraw_amount)
            withdraw_history(account['withdrawal'])
            print(f'''
withdraw_money function:\nbalance: {account['balance']}\nWithdraw Amount: {withdraw_amount}\nwithdraw list: {account['withdrawal']}
            ''')
        else:
            print('your account is not have suffient balance')        
    else:
        print('withdrawal amount must be grater than 0')

## CHECK BALANCE
def check_balance(account):
    current_balance = account['balance']
    print(f'''Your current balance: {current_balance}''')


## PRINT STATEMENT
def print_statement():
    with open(deposit_path, 'r') as file:
        data = json.load(file)
        print(f'statement data: {data}')

# create_account('Nabeel', 0, [])
# deposit_money(account, 2100)
# withdraw_money(account, 20)
# check_balance(account)
# print_statement()
```
