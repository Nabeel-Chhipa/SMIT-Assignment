```python
class Person:
    def __init__(self, name, age, country):
        self.name = name
        self.age = age
        self.country = country
        
    def display(self):
        return f'Name: {self.name}, Age: {self.age}, Country: {self.country}'

p1 = Person('Nabeel', 27, 'Pakistan')
p2 = Person('Usama', 26, 'India')
p1.display()
# p2.display()
```

```python
class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
    
    def display(self):
        return f"Vehicle Details => Make: {self.make}, Model: {self.model}, Year: {self.year}"

class Car(Vehicle):
    def __init__(self, make, model, year, doors):
        super().__init__(make, model, year)
        self.doors = doors

    def display(self):
        return f"Car Details => Make: {self.make}, Model: {self.model}, Year: {self.year}, Doors: {self.doors}"

v1 = Vehicle('Honda', 'Civic', 2024)
c1 = Car('Toyota', 'Yaris', 2024, 2)
v1.display()
# c1.display()
```

```python
class BankAccount:
    def __init__(self, account_number, balance=0):
        self.account_number = account_number
        self.balance = balance

    def deposit(self, deposit_amount=0):
        self.balance += deposit_amount
        return f"Deposit Amount {self.balance}"

    def withdraw(self, withdraw_amount=0):
        self.balance -= withdraw_amount
        return f"Wthdraw Amount {self.balance}"

user1 = BankAccount(123)
user1.deposit(1)
user1.deposit(2)
user1.deposit(5)
user1.withdraw(2)
        
```

## 5

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        PI = 3.142
        calculate = PI * self.radius * self.radius
        return f"Circle Area: {calculate}"

class Triangle(Shape):
    def __init__(self, height, width):
        self.height = height
        self.width = width
    def area(self):
        calculate = 0.5 * self.width * self.height
        return f"Triangle Area: {calculate}"

circle = Circle(5)
triangle = Triangle(10, 5)
circle.area()
triangle.area()
```

```python
from abc import ABC, abstractmethod

class Employee(ABC):
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
    def calculate_annual_salary(self):
        pass

class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)
        self.department = department
    def calculate_annual_salary(self):
        base = 0.10 * self.salary
        total_salary = base + self.salary
        return f"Manager: {self.name}, Department: {self.department}, salary after bonus: {total_salary}"

manager1 = Manager('Nabeel', 200000, 'IT')
manager2 = Manager('Usama', 50000, 'Finance')
manager1.calculate_annual_salary()
manager2.calculate_annual_salary()
```
