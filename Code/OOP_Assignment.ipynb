```python
class Person:
    def __init__(self, name, age, country):
        self.name = name
        self.age = age
        self.country = country
        
    def display(self):
        return f'Name: {self.name}, Age: {self.age}, Country: {self.country}'

p1 = Person('Nabeel', 27, 'Pakistan')
p2 = Person('Usama', 26, 'India')
p1.display()
# p2.display()
```

```python
class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
    
    def display(self):
        return f"Vehicle Details => Make: {self.make}, Model: {self.model}, Year: {self.year}"

class Car(Vehicle):
    def __init__(self, make, model, year, doors):
        super().__init__(make, model, year)
        self.doors = doors

    def display(self):
        return f"Car Details => Make: {self.make}, Model: {self.model}, Year: {self.year}, Doors: {self.doors}"

v1 = Vehicle('Honda', 'Civic', 2024)
c1 = Car('Toyota', 'Yaris', 2024, 2)
v1.display()
# c1.display()
```

```python
class BankAccount:
    def __init__(self, account_number, balance=0):
        self.account_number = account_number
        self.balance = balance

    def deposit(self, deposit_amount=0):
        self.balance += deposit_amount
        return f"Deposit Amount {self.balance}"

    def withdraw(self, withdraw_amount=0):
        self.balance -= withdraw_amount
        return f"Wthdraw Amount {self.balance}"

user1 = BankAccount(123)
user1.deposit(1)
user1.deposit(2)
user1.deposit(5)
user1.withdraw(2)
        
```
