## 2

```python
user_data = input('Enter your data')
upper_case = 0
lower_case = 0
digits = 0
white_space = 0
for char in user_data:
    if char.isupper():
        upper_case += 1
    elif char.islower():
        lower_case += 1
    elif char.isdecimal():
        digits += 1
    elif char.isspace():
        white_space += 1

print(f'Number of upper case letter {upper_case}')
print(f'Number of lower case letter {lower_case}')
print(f'Number of digits {digits}')
print(f'Number of white space in string {white_space}')
```

# 9

```python
password = input('Enter Your Password')
if len(password) < 8:
    print('The password must be at least eight characters long')
if any(char.isupper() for char in password):
    pass
else:
    print('at least one uppercase letter is required')
if any(char.islower() for char in password):
    pass
else:
    print('at least one lowercase letter is required')
if any(char.isdigit() for char in password):
    pass
else:
    print('at least one numeric digit is required')
```
